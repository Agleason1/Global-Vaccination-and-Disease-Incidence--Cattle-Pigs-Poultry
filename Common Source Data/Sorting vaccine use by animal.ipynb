```python
import pandas as pd
import numpy as np
from collections import Counter
import pycountry
import sys,os
import pickle
```

```python
notebook_dir = os.path.dirname(os.getcwd())
source_data_path=os.path.join(notebook_dir, "Common Source Data")
sys.path.append(source_data_path)
from country_codes import countries
from country_regions import regions
```

```python
cattle_measures=pd.read_csv(os.path.join(source_data_path,'WAHIS data','cattle_control_measures.csv'))
cattle_measures=cattle_measures[cattle_measures['Control measure'].isin(['Official vaccination','Vaccination prohibited'])]

poultry_measures=pd.read_csv(os.path.join(source_data_path,'WAHIS data','poultry_control_measures.csv'))
poultry_measures=poultry_measures[poultry_measures['Control measure'].isin(['Official vaccination','Vaccination prohibited'])]

swine_measures=pd.read_csv(os.path.join(source_data_path,'WAHIS data','swine_control_measures.csv'))
swine_measures=swine_measures[swine_measures['Control measure'].isin(['Official vaccination','Vaccination prohibited'])]

cattle_measures['Animal']=['Cattle']*cattle_measures.shape[0]
poultry_measures['Animal']=['Poultry']*poultry_measures.shape[0]
swine_measures['Animal']=['Swine']*swine_measures.shape[0]

df_animals=pd.concat([cattle_measures,poultry_measures,swine_measures])
df_animals=df_animals.sort_values(['Year','Semester'],ascending=[True, True]).drop_duplicates(['Country','Disease','Animal'],keep='last')
```

```python
counts_cattle=Counter(df_animals[(df_animals['Animal']=='Cattle') & (df_animals['Control measure']=='Official vaccination')]['Disease'].values)
counts_cattle=counts_cattle.most_common()
counts_cattle=[('Cattle',i[0],i[1]) for i in counts_cattle]
```

```python
counts_poultry=Counter(df_animals[(df_animals['Animal']=='Poultry') & (df_animals['Control measure']=='Official vaccination')]['Disease'].values)
counts_poultry=counts_poultry.most_common()
counts_poultry=[('Poultry',i[0],i[1]) for i in counts_poultry]
```

```python
counts_swine=Counter(df_animals[(df_animals['Animal']=='Swine') & (df_animals['Control measure']=='Official vaccination')]['Disease'].values)
counts_swine=counts_swine.most_common()
counts_swine=[('Swine',i[0],i[1]) for i in counts_swine]
```

```python
counts_animals=counts_cattle+counts_poultry+counts_swine

counts_animals=sorted(
    counts_animals, 
    key=lambda x: x[2],
    reverse=True
    )
counts_animals
```

```python
with open(os.path.join(source_data_path, 'Processed Data', 'WAHIS data','vaccines_sorted_by_use.pickle'), 'wb') as handle:
    pickle.dump([counts_cattle,counts_poultry,counts_swine,counts_animals], handle, protocol=pickle.HIGHEST_PROTOCOL)
```
